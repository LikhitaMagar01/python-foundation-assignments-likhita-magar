# .loc and .iloc

.loc
selects using the actual index labels or column names
always starts with 0 same as our list index
df.loc[2:5]
slice includes 2, 3, 4, 5

.iloc
selects data from its actual integer
starts with 1 and remove the last integer row
df.iloc[1:4, [1, 2]]
show 1, 2, 3

```python
# aggregation date to summarize trends in pandas

df = pd.DataFrame(
    {
        "product": ["A", "A", "B", "B", "C"],
        "sales": [100, 150, 200, 250, 300],
        "quantity": [2, 3, 4, 5, 6]
    }
)
```

```python
df['sales'].sum()
```

```python
# calculate slaes for each product
# groupby
df.groupby("product")["sales"].sum()
```

```python
# multiple aggregation

df.groupby("product")["sales"].agg(
    [
        "sum",
        "mean",
        "min",
        "max"
    ]
)
```

```python
df.groupby("product").agg(
    {
        "sales": ["sum", "min", "max"],
        "quantity": "sum"
    }
    
)
```

```python
df.groupby("product").agg(
    total_sales = ("sales", "sum"),
    average_sales = ("sales", "min")
)
```

```python
df.groupby("product").size()
```

```python
df.groupby('product')['sales'].count()
```
