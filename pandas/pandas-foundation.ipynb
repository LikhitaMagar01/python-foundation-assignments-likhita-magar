# .loc and .iloc

.loc
selects using the actual index labels or column names
always starts with 0 same as our list index
df.loc[2:5]
slice includes 2, 3, 4, 5

.iloc
selects data from its actual integer
starts with 1 and remove the last integer row
df.iloc[1:4, [1, 2]]
show 1, 2, 3

```python
df.loc[0:4, ["name", "city"]]
```

```python
df.iloc[0:3, [1, 2]]
```

```python
df.loc[0:2, ["name", "city"]]
```
