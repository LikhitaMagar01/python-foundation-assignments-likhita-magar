```python
# vectorized operation
# applies the same calculations to every element at once


# why numpy is much faster
# - no need to check datatype
# - can add variable without list loop

# Broadcasting

import time
import numpy as np

# Create a list and an array with 10 million elements
size = 10_000_000
python_list = list(range(size))
numpy_array = np.arange(size)

# 1. Pure Python Loop
start_time = time.time()
python_list_result = [x + 1 for x in python_list]
python_time = time.time() - start_time
print(f"Pure Python Time: {python_time:.5f} seconds")

# 2. Vectorized NumPy Operation
start_time = time.time()
numpy_array_result = numpy_array + 1
numpy_time = time.time() - start_time
print(f"NumPy Time: {numpy_time:.5f} seconds")

# Speedup factor
print(f"NumPy is {python_time / numpy_time:.1f}x faster!")
```
