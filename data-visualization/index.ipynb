```python
import matplotlib.pyplot as plt 
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
districts = ["ktm", "lalitpur", "bhaktapur", "kaski", "rupendehi"]
sales = [26, 45, 10, 12, 14]
```

```python
# set figure

plt.figure(figsize=(5,3)) # (width, height) -> inch
plt.bar(districts, sales, color="#2F5D8A")
plt.title("Sales by District")
plt.xlabel("Districts")
plt.ylabel("Sales")
plt.show()
```

```python
# plot sin wave

import numpy as np 

x = np.linspace(0, 10, 50)
y = np.sin(x)
z = np.cos(x)

print(x)
print(y)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='blue', label="Sin(x)")
ax.plot(x, z, color='red', label="Sin(z)")
ax.set_title("Sine wave - plot")
ax.set_xlabel("x")
ax.set_ylabel("sign(x)")
#plt.show()
```

```python

# subplots with multiple axis:


fig, axes = plt.subplots(1, 3, figsize=(12, 3))

axes[0].plot(x, y)
axes[0].set_title("sine plot")

axes[1].bar(districts, sales)
axes[1].set_title("Sales by District")

axes[2].plot(x, z)
axes[2].set_title("cos plot")

plt.tight_layout()

plt.show()
```

```python
plt.savefig("multiple_plot.png")
```

```python
# statistical visualization with seaborn

tips = sns.load_dataset("tips")

# scatter plot
fig, ax = plt.subplots(figsize=(5,3))
for day, group in tips.groupby("day", observed=True):
    ax.scatter(group['total_bill'], group['tip'], label=day)
ax.set_title("Scatter plot")
ax.set_xlabel("Total Bill")
ax.set_ylabel("Tip")
plt.show()
```

```python
sns.relplot(data=tips, x="total_bill", y="tip", hue="day", height=4, aspect=1.5)
plt.show()
```

```python
# scatter plot: total bill vs size

sns.scatterplot(data=tips, x="total_bill", y="size", hue="time", sizes=(70, 500), marker="x")
plt.show()
```

```python
sns.histplot(data=tips, x="total_bill", kde=True)
plt.title("Distribution: histplot")
plt.show()
```

```python
sns.boxplot(data=tips, x="day", y="total_bill")
plt.title("Categorical: boxplot")
plt.show()
```

```python
sns.heatmap(tips.corr(numeric_only=True), annot=True, cmap="Blues")
plt.title("Matrix: heatmap")
plt.show()
```
